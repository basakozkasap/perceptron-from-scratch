```python

# PERCEPTRON

# 1. Initial parameters

input_count = int(input("Enter number of inputs: "))

weights = []

for i in range(input_count):
    weight = float(input(f"Enter weight {i + 1}: "))
    weights.append(weight)

alpha = float(input("Enter learning rate: "))
bias = float(input("Enter bias: "))
phi = float(input("Enter threshold (phi): "))


# 2. Training data

training_data = []

sample_count = int(input("Enter number of samples: "))

for i in range(sample_count):
    print(f"\nSample {i + 1}")

    inputs = []

    for j in range(input_count):
        x = float(input(f"Enter x{j + 1}: "))
        inputs.append(x)

    target = int(input("Enter target (-1 or 1): "))

    training_data.append((inputs, target))


# 3. Perceptron training

epoch = 0
max_epochs = 1000

while epoch < max_epochs:
    epoch += 1
    has_error = False

    for inputs, target in training_data:

        # Calculate net input
        net = bias - phi

        for j in range(input_count):
            net += weights[j] * inputs[j]

        # Activation function
        if net >= 0:
            output = 1
        else:
            output = -1

        # Calculate error
        error = target - output

        # Update parameters
        if error != 0:
            has_error = True

            for j in range(input_count):
                weights[j] += alpha * error * inputs[j]

            bias += alpha * error
            phi -= alpha * error

    # Stop when all predictions are correct
    if not has_error:
        break


# 4. Final results

print("\n===== TRAINING RESULTS =====")

if not has_error:
    print("Training completed successfully!")
else:
    print("Maximum epoch limit reached.")

print("Epochs:", epoch)

for i in range(input_count):
    print(f"Weight {i + 1}:", round(weights[i], 4))

print("Bias:", round(bias, 4))
print("Threshold (phi):", round(phi, 4))


# 5. Final predictions

print("\n===== FINAL PREDICTIONS =====")

for inputs, target in training_data:

    net = bias - phi

    for j in range(input_count):
        net += weights[j] * inputs[j]

    output = 1 if net >= 0 else -1

    print(
        f"Inputs: {inputs} | "
        f"Target: {target} | "
        f"Output: {output}"
    )
```
